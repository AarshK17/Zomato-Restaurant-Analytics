# Does online ordering relate to restaurant ratings? 

### Rate(numeric), Online Ordering(categorical) - Bivariate

```python
df
```

```python
plt.figure(figsize = (7,5))
sns.boxplot(
    data = df,
    x = 'online_order',
    y = 'rate'
)
plt.xlabel("Online Ordering")
plt.ylabel("Restaurent Rating")
plt.title("Restaurent Rating by Online Ordering Availability")
plt.show()
```

# # The Yes group has a somewhat higher median, but there is substantial overlap.

```python
df.groupby('online_order')['rate'].agg(['count', 'mean', 'median', 'min', 'max'])
```

### Restaurants offering online ordering have slightly higher average and median ratings than restaurants without online ordering (3.72 vs. 3.66 mean; 3.8 vs. 3.7 median). However, the difference is modest and the rating distributions overlap considerably, so this analysis shows an association rather than evidence that online ordering causes higher ratings.

# Is restaurant price associated with restaurant rating?
### approx_cost(numerical), rate(numerical) - Scatter Plot Analysis

```python
price_rating = df[['approx_cost(for two people)', 'rate']].dropna()
print(price_rating.shape)
```

```python
plt.figure(figsize = (8,5))
sns.scatterplot(
    data = price_rating,
    x = 'approx_cost(for two people)',
    y = 'rate',
    alpha = 0.3
)
plt.xlabel("Approx Cost For Two People")
plt.ylabel("Restaurent Rating")
plt.title("Prive vs Rating")
plt.show()
```

```python
price_rating['approx_cost(for two people)'].corr(price_rating['rate'])
```

### Restaurant price shows a weak-to-moderate positive relationship with ratings (Pearson correlation = 0.385). Higher-priced restaurants tend to have somewhat higher ratings, but the substantial spread of ratings at different price levels indicates that price alone is not a strong predictor of restaurant ratings.

```python
top_cuisines = cuisine_counts.head(10)
plt.figure(figsize = (10, 6))
sns.barplot(
    x = top_cuisines.values,
    y = top_cuisines.index
)
plt.xlabel("Number of restaurent Listings")
plt.ylabel('Cuisine')
plt.title('Top 10 Most common Cuisine')
plt.show()
```

### which cuisine recieve higher rating ? 

```python
cuisine_rating = df[['cuisines', 'rate']].dropna(subset = ['rate']).copy()
cuisine_rating['cuisines'] = cuisine_rating['cuisines'].str.split(',')
cuisine_rating = cuisine_rating.explode('cuisines')
cuisine_rating['cuisines'] = cuisine_rating['cuisines'].str.strip()
```

```python
cuisine_rating_stats = (
    cuisine_rating
    .groupby('cuisines')['rate']
    .agg(['count', 'mean', 'median'])
    .sort_values('count', ascending = False)
)
cuisine_rating_stats.head(15)
```

### Restaurants offering Continental and Italian cuisine have relatively high average ratings in this dataset, while restaurants offering Biryani, Chinese, South Indian and Fast Food have lower average ratings.

```python
cuisine_rating_filtered = cuisine_rating_stats[
    cuisine_rating_stats['count'] >= 1000
].sort_values('mean', ascending=False)

cuisine_rating_filtered
```

```python
plt.figure(figsize=(10, 7))
sns.barplot(
    data=cuisine_rating_filtered.reset_index(),
    x='mean',
    y='cuisines'
)
plt.xlabel('Average Rating')
plt.ylabel('Cuisine')
plt.title('Average Rating by Cuisine')
plt.show()
```

# Does online ordering and table booking relate to restaurant ratings?

```python
service_rating = (
    df.groupby(['online_order', 'book_table'])['rate']
      .agg(['count', 'mean', 'median'])
      .reset_index()
)
service_rating
```

### Restaurants offering table booking show substantially higher average ratings than those without table booking. The association is consistent across both online-ordering groups, while online ordering alone shows a much smaller difference.  

```python
plt.figure(figsize=(9, 6))
sns.barplot(
    data=service_rating,
    x='online_order',
    y='mean',
    hue='book_table'
)
plt.xlabel('Online Ordering')
plt.ylabel('Average Rating')
plt.title('Average Restaurant Rating by Online Ordering and Table Booking')
plt.ylim(0, 5)
plt.show()
```

# What is the typical price range for two people, and how widely do restaurant prices vary?

```python
df['approx_cost(for two people)'].describe()
```

###  Because the mean is considerably higher than the median, this suggests that the price distribution is right-skewed a relatively small number of expensive restaurants are pulling the average upward.

```python
plt.figure(figsize=(9, 6))
sns.histplot(
    df['approx_cost(for two people)'].dropna(),
    bins=30
)
plt.xlabel('Approximate Cost for Two People (₹)')
plt.ylabel('Number of Restaurants')
plt.title('Distribution of Restaurant Prices')
plt.show()
```

```python
counts, bins = np.histogram(
    df['approx_cost(for two people)'].dropna(),
    bins=30
)

for i in range(len(counts)):
    print(f"₹{bins[i]:.0f} - ₹{bins[i+1]:.0f}: {counts[i]}")
```

### Restaurant prices are strongly right-skewed. The median approximate cost for two people is ₹400, compared with a mean of ₹555, with most restaurants concentrated below ₹650. A small number of high-priced restaurants extend the distribution up to ₹6,000.

# Do restaurant ratings vary with price and customer engagement (votes)?

```python
rating_price_votes = df[
    ['rate', 'approx_cost(for two people)', 'votes']
].dropna()

print(rating_price_votes.shape)
```

```python
rating_price_votes.corr()
```

### Restaurant ratings show moderate positive associations with both customer votes (r = 0.435) and approximate cost (r = 0.385). Customer votes have a slightly stronger correlation with ratings than price, while price and votes are also moderately correlated (r = 0.367). These relationships indicate that price and customer engagement are associated with ratings, but neither variable alone strongly explains restaurant ratings.
